```python
from pathlib import Path
import pandas as pd

DENGUE_PATH = Path("../../dengue")

if not DENGUE_PATH.exists():
    raise FileNotFoundError(f"Directory not found: {DENGUE_PATH.absolute()}")

csv_files = list(DENGUE_PATH.glob("Datos_*_210.csv"))
print(f"Found {len(csv_files)} file(s) in: {DENGUE_PATH.absolute()}")

data = []
for file in csv_files:
    try:
        year = int(file.stem.split("_")[1])
        df = pd.read_csv(file, encoding="latin-1", on_bad_lines="skip")
        data.append({"year": year, "count": len(df)})
    except Exception as e:
        print(f"Skipped {file.name}: {e}")

if data:
    result_df = pd.DataFrame(data).sort_values("year").reset_index(drop=True)
else:
    result_df = pd.DataFrame(columns=["year", "count"])
    print("Warning: No valid data processed.")

output_path = DENGUE_PATH / "dengue_yearly_counts.csv"
result_df.to_csv(output_path, index=False)
print(f"Successfully created: {output_path}")
```
