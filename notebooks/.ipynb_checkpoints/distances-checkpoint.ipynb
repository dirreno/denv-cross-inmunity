```python
def edit_distance(s1, s2):
    if len(s1) != len(s2):
        raise ValueError("Strings must be of the same length")
    
    return sum(el1 != el2 for el1, el2 in zip(s1, s2))
```

```python
def load_or_create_db(gff_path, db_name):
    
    if not os.path.exists(db_name):
        print(f"Creating {db_name}...")
        gffutils.create_db(
            gff_path,
            dbfn=db_name,
            force=True,
            keep_order=True,
            merge_strategy="merge",
            sort_attribute_values=True
        )
    else:
        print(f"{db_name} already exists. Loading...")

    return gffutils.FeatureDB(db_name)


db_denv1 = load_or_create_db("../resources/test/meta_ref_DENV1.gff3", "denv1.db")
db_denv2 = load_or_create_db("../resources/test/meta_ref_DENV2.gff3", "denv2.db")
db_denv3 = load_or_create_db("../resources/test/meta_ref_DENV3.gff3", "denv3.db")
db_denv4 = load_or_create_db("../resources/test/meta_ref_DENV4.gff3", "denv4.db")
```

```python
def get_feature_coords(db, feature_name):
    for f in db.features_of_type("mature_protein_region_of_CDS"):
        product_name = f.attributes.get("product", [""])[0]
        if feature_name in product_name:
            return f.start, f.end, f.strand
    return None
```

## Augur Functions

```python
def augur_align(in_seqs=None, ref_seq=None, out_aln=None, fill_gaps=False, nthreads="auto"):
    cmd = [
        "augur", "align",
        "--sequences", in_seqs,
        "--output", out_aln,
        "--nthreads", str(nthreads)
    ]
    if ref_seq:
        cmd.extend(["--reference-sequence", ref_seq])
    if fill_gaps:
        cmd.append("--fill-gaps")
    subprocess.run(cmd, check=True)
```

```python
def augur_tree(in_seqs=None, method="iqtree", sub_model="GTR+G", nthreads="auto", out_tree=None):
    subprocess.run([
            "augur", "tree",
            "--alignment", in_seqs,
            "--method", method,
            "--substitution-model", sub_model,
            "--nthreads", nthreads,
            "--output", out_tree
        ], check=True)
```

```python
def augur_refine(in_tree=None, in_seqs=None, in_meta=None, clock="4", out_tree=None, out_node=None):
    subprocess.run([
            "augur", "refine",
            "--tree", in_tree,
            "--alignment", in_seqs,
            "--metadata", in_meta,
            "--timetree",
            "--clock-filter-iqd", clock,
            "--output-tree", out_tree,
            "--output-node-data", out_node,
            "--stochastic-resolve"
        ], check=True)
```

```python
def augur_ancestral(in_tree=None, in_seqs=None, out_node=None, out_seqs=None):
    subprocess.run([
        "augur", "ancestral",
        "--tree", in_tree,
        "--alignment", in_seqs,
        "--output-node-data", out_node,
        "--output-sequences", out_seqs
    ], check=True)
```

```python
def augur_translate(in_tree=None, ancestral_seqs=None, ref_seq=None, genes=None, out_node=None, out_aln=None):
    cmd = [
        "augur", "translate",
        "--tree", in_tree,
        "--ancestral-sequences", ancestral_seqs,
        "--reference-sequence", ref_seq,
        "--output-node-data", out_node,
        "--alignment-output", out_aln
    ]
    if genes:
        cmd.append("--genes")
        cmd.extend(genes if isinstance(genes, list) else [genes])
    subprocess.run(cmd, check=True)
```

```python
IN_SEQS_COL = "../resources/colombia_data/sequences/col_DENV{}.fasta"
REF_SEQ = "../resources/reference_data/sequences/ref_DENV{}.fasta"
ALN_SEQS = "../resources/colombia_data/aligned/align_DENV{}.fasta"
CROP_SEQS = "../resources/colombia_data/cropped/align_DENV{}.fasta"
META_COL = "../resources/colombia_data/metadata/col_DENV{}_meta.tsv"
META_CLEAN_COL = "../resources/colombia_data/metadata/E_DENV{}_meta.tsv"
TMP_GFF = "../resources/colombia_data/tmp/E_DENV{}.gff"
TREE_COL = "../resources/colombia_data/tree/E_DENV{}_tree.nwk"
TIME_TREE_COL = "../resources/colombia_data/refined/E_DENV{}_tree_time.nwk"
NODE_TREE_COL = "../resources/colombia_data/refined/E_DENV{}_tree_node.json"
NODE_AA_COL="../resources/colombia_data/AA_sequences/E_DENV{}_aa.json"
AA_SEQS_COL="../resources/colombia_data/AA_sequences/E_DENV{}_aa_%GENE.fasta"
AA_SEQS_COL_E="../resources/colombia_data/AA_sequences/E_DENV{}_aa_E.fasta"
ANCESTRAL_NODE="../resources/colombia_data/ancestral/E_DENV{}_ancestral.json"
ANCESTRAL_SEQS="../resources/colombia_data/ancestral/E_DENV{}_ancestral.fasta"
```

## Align sequences

```python
for i in range(4):
    print(f"Aligning DENV{i+1}")
    serotype = i + 1
    augur_align(
        in_seqs=IN_SEQS_COL.format(serotype),
        ref_seq=REF_SEQ.format(serotype),
        out_aln=ALN_SEQS.format(serotype),
        fill_gaps=True,
        nthreads="auto"
    )
```

## Crop alignments gene E

```python
for i in range(4):
    print(f"Crop DENV{i+1}")
    serotype = i + 1
    db = [db_denv1, db_denv2, db_denv3, db_denv4][i]
    e_start, e_end, strand = get_feature_coords(db, "envelope protein E")
    cropped_records = []
    
    for j, record in enumerate(SeqIO.parse(ALN_SEQS.format(serotype), "fasta")):
        
        if j == 0:
            continue
            
        if len(record.seq) < e_end:
            continue
        
        cropped_seq = record.seq[e_start-1:e_end]
        
        if strand == "-":
            cropped_seq = cropped_seq.reverse_complement()
        
        record.seq = cropped_seq

        new_id = record.id.split("|")[0]
        
        record.id = new_id
        record.name = new_id
        record.description = new_id
        
        cropped_records.append(record)
    
    SeqIO.write(
        cropped_records,
        CROP_SEQS.format(serotype),
        "fasta"
    )
```

## Clean Metadata

```python
import pandas as pd

for i in range(4):
    print(f"Clean Metadata DENV{i+1}")
    serotype = i + 1
    
    df = pd.read_csv(META_COL.format(serotype), sep="\t")
    
    df = df.rename(columns={
        "Virus name": "strain",
        "Collection date": "date",
        "Location": "location"
    })
    
    location_split = df["location"].str.split(" / ", expand=True)
    
    df["region"] = location_split[0]
    df["country"] = location_split[1]
    df["division"] = location_split[2]
    
    metadata = df[["strain", "date", "region", "country", "division"]]
   
    metadata.to_csv(META_CLEAN_COL.format(serotype), sep="\t", index=False)
```

## Create temp GFF

```python
for i in range(4):
    print(f"Create temporal GFF DENV{i+1}")
    serotype = i + 1
    
    first_record = next(SeqIO.parse(CROP_SEQS.format(serotype), "fasta"))
    gene_length = len(first_record.seq)
    seqid = first_record.id
    
    with open(TMP_GFF.format(serotype), "w") as gff:
        gff.write("##gff-version 3\n")
        
        gff.write(
            f"{seqid}\t.\tsource\t1\t{gene_length}\t.\t+\t.\tID=nuc\n"
        )
        
        gff.write(
            f"{seqid}\t.\tgene\t1\t{gene_length}\t.\t+\t.\tID=E;Name=E;gene=E\n"
        )
```

## Calculate Distances

```python
sns.set_theme(style="whitegrid", context="notebook")

plot_data = []
all_years = None

# First pass: collect data and determine global year range
for sero in range(1, 5):
    df = pd.read_csv(META_CLEAN_COL.format(sero), sep="\t")
    df["date"] = pd.to_datetime(df["date"], format="mixed", errors="coerce")
    df = df.dropna(subset=["date"])
    df["year"] = df["date"].dt.year
    
    counts = df.groupby("year").size()
    
    # Track global min/max year
    if all_years is None:
        all_years = range(counts.index.min(), counts.index.max() + 1)
    else:
        all_years = range(min(all_years[0], counts.index.min()), 
                         max(all_years[-1], counts.index.max()) + 1)
    
    # Reindex to fill missing years with 0
    counts = counts.reindex(all_years, fill_value=0).reset_index()
    counts.columns = ["year", "count"]
    counts["serotype"] = f"DENV{sero}"
    plot_data.append(counts)

# Combine all serotypes
plot_df = pd.concat(plot_data, ignore_index=True)

# Plot
plt.figure(figsize=(10, 6))
sns.lineplot(
    data=plot_df,
    x="year",
    y="count",
    hue="serotype",
    marker="o",
    linewidth=2,
    palette="husl"
)

plt.xlabel("Año", fontsize=12)
plt.ylabel("Número de secuencias", fontsize=12)
plt.title("Número de secuencias del gen E", fontsize=14, fontweight="bold")
plt.legend(title="Serotipo")
plt.xticks(plot_df["year"].unique().astype(int), rotation=45, fontsize=6)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from Bio import SeqIO
from matplotlib.ticker import MaxNLocator

# Configurar estilo de Seaborn
sns.set_theme(style="whitegrid", context="notebook", palette="husl")

plot_data = []
year_ranges = []

# PRIMER PASO: Cargar datos y recopilar rangos de años de cada serotipo
for s in range(4):
    sero = s + 1
    
    # Cargar secuencias
    records = list(SeqIO.parse(AA_SEQS_COL_E.format(sero), "fasta"))
    ids = [r.id for r in records]
    seqs = [str(r.seq) for r in records]
    n = len(seqs)
    
    # Calcular matriz de distancias
    matrix = np.zeros((n, n))
    for i in range(n):
        for j in range(i, n):
            dist, sites = grantham_distance(seqs[i], seqs[j])
            matrix[i, j] = dist
            matrix[j, i] = dist
    
    df_matrix = pd.DataFrame(matrix, index=ids, columns=ids)
    
    # Cargar metadata
    meta = pd.read_csv(META_CLEAN_COL.format(sero), sep="\t")
    meta["date"] = pd.to_datetime(meta["date"], format="mixed", errors="coerce")
    meta = meta.dropna(subset=["date"])
    meta["year"] = meta["date"].dt.year
    
    # Convertir a formato largo y fusionar con metadata
    df_long = df_matrix.stack().reset_index()
    df_long.columns = ["seq1", "seq2", "distance"]
    
    df_long = df_long.merge(meta[["strain", "year"]], left_on="seq1", right_on="strain")
    df_long = df_long.rename(columns={"year": "year1"}).drop(columns=["strain"])
    
    df_long = df_long.merge(meta[["strain", "year"]], left_on="seq2", right_on="strain")
    df_long = df_long.rename(columns={"year": "year2"}).drop(columns=["strain"])
    
    # Promedio de años para el par
    df_long["year"] = ((df_long["year1"] + df_long["year2"]) / 2).astype(int)
    df_long = df_long[df_long["seq1"] != df_long["seq2"]]
    
    # Calcular promedio anual
    avg = df_long.groupby("year")["distance"].mean()
    
    # Guardar datos y rango de años
    plot_data.append({"sero": sero, "avg": avg})
    year_ranges.append((avg.index.min(), avg.index.max()))

# Determinar el rango global de años verdadero entre todos los serotipos
global_min = min(r[0] for r in year_ranges)
global_max = max(r[1] for r in year_ranges)
global_years = range(global_min, global_max + 1)

# SEGUNDO PASO: Reindexar cada serotipo al rango global completo
final_data = []
for item in plot_data:
    # Reindexar con NaN para crear gaps honestos en lugar de conectar puntos discontinuos
    avg_smooth = item["avg"].reindex(global_years, fill_value=np.nan)
    df_temp = avg_smooth.reset_index().rename(columns={"index": "year", "distance": "avg_distance"})
    df_temp["serotype"] = f"DENV{item['sero']}"
    final_data.append(df_temp)

plot_df = pd.concat(final_data, ignore_index=True)

# Crear el plot con Seaborn
plt.figure(figsize=(12, 7))
sns.lineplot(
    data=plot_df,
    x="year",
    y="avg_distance",
    hue="serotype",
    marker="o",
    linewidth=2.5,
    markersize=5,
    palette="husl"
)

# Etiquetas y título
plt.xlabel("Año", fontsize=12, fontweight="bold")
plt.ylabel("Distancia Grantham promedio", fontsize=12, fontweight="bold")
plt.title("Cambio evolutivo en el tiempo (gen E) - DENV1–4", fontsize=14, fontweight="bold", pad=20)

# Forzar xticks anuales enteros y legibles
plt.xticks(ticks=list(global_years), labels=list(global_years), rotation=45, ha="right")

# Leyenda y estilo
plt.legend(title="Serotipo", title_fontsize=11, fontsize=10)
plt.grid(axis="y", alpha=0.3)
sns.despine(top=True, right=True)

plt.tight_layout()
plt.show()
```

## CROP ALIGMENTES GENE E & CREATE METADATA

```python
from Bio import SeqIO
import subprocess
import pandas as pd
import os
from collections import defaultdict


for i in range(4):
    sero = i+1
    
    db = [db_denv1, db_denv2, db_denv3, db_denv4][i]
    e_start, e_end, strand = get_feature_coords(db, "envelope protein E")
    
    cropped_records = []
    fasta_in = fastas_aligned[i]
    
    for j, record in enumerate(SeqIO.parse(fasta_in, "fasta")):
        if j == 0: continue
        if len(record.seq) < e_end: continue
            
        cropped_seq = record.seq[e_start-1:e_end]
        if strand == "-":
            cropped_seq = cropped_seq.reverse_complement()
        record.seq = cropped_seq
        cropped_records.append(record)
        
    strain_counts = defaultdict(int)
    augur_ready = []
    
    for record in cropped_records:
        parts = record.description.split()
        parts += ["."] * (5 - len(parts))
        strain = parts[0]
        
        strain_counts[strain] += 1
        if strain_counts[strain] > 1:
            new_strain = f"{strain}_{strain_counts[strain]}"
            strain = new_strain
            
        parts[0] = strain
        record.id = "|".join(parts)
        record.name = strain
        record.description = record.id
        augur_ready.append(record)
        
    SeqIO.write(augur_ready, TMP_CROP_SEQS.format(sero), "fasta")
    meta_df = META_EXP.format(sero)
    subprocess.run([
        "augur", "parse",
        "--sequences", TMP_CROP_SEQS.format(sero),
        "--output-sequences", CROP_SEQS.format(sero),
        "--output-metadata", meta_df,
        "--fields", "strain", "country", "genotype", "accession", "extra",
        "--separator", "|"
    ], check=True)
    
    meta_df = pd.read_csv(meta_tsv, sep="\t")
    meta_df["date"] = meta_df["strain"].str.extract(r"^(\d{4})")[0]
    
    meta_df = meta_df[["strain", "date", "country"]]
    meta_df.to_csv(meta_tsv, sep="\t", index=False)
    
    os.remove(TMP_CROP_SEQS.format(sero))
    print(f"Saved: {final_fasta} & {meta_tsv}")
```

## Create temp GFF

```python
for i in range(4):
    print(f"Create temporal GFF DENV{i+1}")
    serotype = i + 1
    
    first_record = next(SeqIO.parse(CROP_SEQS.format(serotype), "fasta"))
    gene_length = len(first_record.seq)
    seqid = first_record.id
    
    with open(TMP_GFF.format(serotype), "w") as gff:
        gff.write("##gff-version 3\n")
        
        gff.write(
            f"{seqid}\t.\tsource\t1\t{gene_length}\t.\t+\t.\tID=nuc\n"
        )
        
        gff.write(
            f"{seqid}\t.\tgene\t1\t{gene_length}\t.\t+\t.\tID=E;Name=E;gene=E\n"
        )
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from Bio import SeqIO
import re

plt.figure(figsize=(12, 7))

for s in range(4):
    print(f"🔹 Processing DENV{s+1}...")
    fasta_path = f"../resources/AntigenicMaps/Sequences/E_DENV{s+1}_aa_E.fasta"
    records = list(SeqIO.parse(fasta_path, "fasta"))
    
    # 1️⃣ Extract year directly from FASTA ID & skip assembly contigs (NODE_...)
    valid_records = []
    years = []
    for r in records:
        match = re.match(r"^(\d{4})", r.id)
        if match:
            valid_records.append(r)
            years.append(int(match.group(1)))
            
    if len(valid_records) < 2:
        print(f"  ⚠️  Too few sequences with years. Skipping.")
        continue
        
    ids = [r.id for r in valid_records]
    seqs = [str(r.seq) for r in valid_records]
    n = len(seqs)
    print(f"  ✅ {n} sequences with valid years")
    
    # 2️⃣ Compute pairwise Grantham distances
    print(f"  🧮 Computing {n*(n-1)//2} pairwise distances...")
    matrix = np.zeros((n, n))
    for i in range(n):
        for j in range(i+1, n):
            dist, _ = grantham_distance(seqs[i], seqs[j])
            matrix[i, j] = matrix[j, i] = dist
            
    # 3️⃣ Build long-format dataframe
    pairs = [(ids[i], ids[j], years[i], years[j], matrix[i, j]) 
             for i in range(n) for j in range(i+1, n)]
    df = pd.DataFrame(pairs, columns=["id1", "id2", "year1", "year2", "distance"])
    df["year"] = ((df["year1"] + df["year2"]) / 2).astype(int)
    
    # 4️⃣ Plot yearly averages
    avg = df.groupby("year")["distance"].mean()
    plt.plot(avg.index, avg.values, marker='o', label=f"DENV{s+1} (n={len(df)} pairs)", linewidth=2)

plt.xlabel("Año", fontsize=12)
plt.ylabel("Distancia Grantham promedio", fontsize=12)
plt.title("Cambio evolutivo en el tiempo (gen E) - DENV1–4", fontsize=14, fontweight='bold')
plt.legend(fontsize=10)
plt.grid(alpha=0.3, linestyle='--')
plt.tight_layout()
plt.show()
```
