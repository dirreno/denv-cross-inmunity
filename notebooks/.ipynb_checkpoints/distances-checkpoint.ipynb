```python
def hamming_distance(s1, s2):
    if len(s1) != len(s2):
        raise ValueError("Strings must be of the same length")
    
    return sum(el1 != el2 for el1, el2 in zip(s1, s2))
```

### Get Feature Coords

```python
def get_feature_coords(db, feature_name):
    for f in db.features_of_type("mature_protein_region_of_CDS"):
        product_name = f.attributes.get("product", [""])[0]
        if feature_name in product_name:
            return f.start, f.end, f.strand
    return None
```

### Compute Grantham Matrices

```python
def compute_grantham_matrices(
    fasta_path_template: str,
    grantham_func: Callable[[str, str], Tuple[float, int]],
    serotypes: Optional[List[int]] = None,
    ediii_start: int = 297,
    ediii_end: int = 397
) -> List[pd.DataFrame]:
    
    if serotypes is None:
        serotypes = [1, 2, 3, 4]
        
    matrices = []
    
    for sero in serotypes:
        records = list(SeqIO.parse(fasta_path_template.format(sero), "fasta"))
            
        ids = [r.id for r in records]
        seqs = [str(r.seq) for r in records]
        n = len(seqs)
        
        # Pre-cortar región EDIII para mayor eficiencia
        ediii_seqs = [s[ediii_start:ediii_end] for s in seqs]
        
        # Usar NaN en lugar de -1 para compatibilidad con pandas/numpy
        matrix = np.full((n, n), np.nan)
        
        for i in range(n):
            for j in range(i, n):
                dist, sites = grantham_func(ediii_seqs[i], ediii_seqs[j])
                value = dist / sites if sites > 0 else np.nan
                matrix[i, j] = value
                matrix[j, i] = value
                
        df_matrix = pd.DataFrame(matrix, index=ids, columns=ids)
        matrices.append(df_matrix)
        
    return matrices
```

```python
def compute_grantham_distances_vs_reference(
    fasta_path_template: str,
    ref_path_template: str,
    meta_path_template: str,
    grantham_func: Callable[[str, str], Tuple[float, int]],
    serotypes: Optional[List[int]] = None
) -> pd.DataFrame:
    
    if serotypes is None:
        serotypes = [1, 2, 3, 4]
        
    all_data = []
    
    for sero in serotypes:
        records = list(SeqIO.parse(fasta_path_template.format(sero), "fasta"))
            
        ids = [r.id.split('|')[0] if '|' in r.id else r.id for r in records]
        seqs = [str(r.seq) for r in records]

        ref_records = list(SeqIO.parse(ref_path_template.format(sero), "fasta"))
        ref_seq = str(ref_records[0].seq)
        
        distances = []
        for seq in seqs:
            dist, sites = grantham_func(seq, ref_seq)
            distances.append(dist / sites if sites > 0 else np.nan)
            
        df_dist = pd.DataFrame({"strain": ids, "distance": distances}).dropna(subset=["distance"])
            
        meta = pd.read_csv(meta_path_template.format(sero), sep="\t")
        meta["date"] = pd.to_datetime(meta["date"], format="mixed", errors="coerce")
        meta = meta.dropna(subset=["date"]).copy()
        meta["year"] = meta["date"].dt.year
            
        df_combined = df_dist.merge(meta[["strain", "year"]], on="strain", how="inner")
            
        df_combined["serotype"] = sero
        all_data.append(df_combined[["serotype", "strain", "distance", "year"]])
        
    return pd.concat(all_data, ignore_index=True)
```

```python
def compute_grantham_cross_matrices_for_plotter(
    fasta_col_template: str,
    fasta_exp_template: str,
    meta_col_template: str,
    meta_exp_template: str,
    grantham_func: Callable[[str, str], Tuple[float, int]],
    serotypes: Optional[List[int]] = None,
    region_slice: Optional[Tuple[int, int]] = (297, 397),
    exclude_prefix: str = "NODE",
    output_meta_dir: str = "merged_metadata"
) -> Tuple[List[pd.DataFrame], str]:
    
    if serotypes is None:
        serotypes = [1, 2, 3, 4]
        
    os.makedirs(output_meta_dir, exist_ok=True)
    matrices = []
    
    for sero in serotypes:
        records_col = [r for r in SeqIO.parse(fasta_col_template.format(sero), "fasta") 
                      if not r.id.startswith(exclude_prefix)]
        records_exp = [r for r in SeqIO.parse(fasta_exp_template.format(sero), "fasta") 
                      if not r.id.startswith(exclude_prefix)]
            
        ids_col = [r.id for r in records_col]
        seqs_col = [str(r.seq) for r in records_col]
        ids_exp = [r.id for r in records_exp]
        seqs_exp = [str(r.seq) for r in records_exp]
        
        if region_slice:
            start, end = region_slice
            seqs_col = [s[start:end] for s in seqs_col]
            seqs_exp = [s[start:end] for s in seqs_exp]
            
        n, m = len(seqs_col), len(seqs_exp)
        matrix = np.full((n, m), np.nan)
        for i in range(n):
            for j in range(m):
                dist, sites = grantham_func(seqs_col[i], seqs_exp[j])
                matrix[i, j] = dist / sites if sites > 0 else np.nan
                
        df_matrix = pd.DataFrame(matrix, index=ids_col, columns=ids_exp)
        matrices.append(df_matrix)
        
        meta_col = pd.read_csv(meta_col_template.format(sero), sep="\t")
        meta_exp = pd.read_csv(meta_exp_template.format(sero), sep="\t")

        for df in [meta_col, meta_exp]:
            if "strain" not in df.columns:
                df["strain"] = df.iloc[:, 0]
                
        meta_merged = pd.concat([meta_col, meta_exp], ignore_index=True).drop_duplicates(subset=["strain"])
        merged_path = f"{output_meta_dir}/meta_merged_{sero}.tsv"
        meta_merged.to_csv(merged_path, sep="\t", index=False)
        
    merged_template = f"{output_meta_dir}/meta_merged_{{}}.tsv"
    return matrices, merged_template
```

## Augur Functions

```python
def augur_align(in_seqs=None, ref_seq=None, out_aln=None, fill_gaps=False, nthreads="auto"):
    cmd = [
        "augur", "align",
        "--sequences", in_seqs,
        "--output", out_aln,
        "--nthreads", str(nthreads)
    ]
    if ref_seq:
        cmd.extend(["--reference-sequence", ref_seq])
    if fill_gaps:
        cmd.append("--fill-gaps")
    subprocess.run(cmd, check=True)
```

```python
def augur_tree(in_seqs=None, method="iqtree", sub_model="GTR+G", nthreads="auto", out_tree=None):
    subprocess.run([
            "augur", "tree",
            "--alignment", in_seqs,
            "--method", method,
            "--substitution-model", sub_model,
            "--nthreads", nthreads,
            "--output", out_tree
        ], check=True)
```

```python
def augur_refine(in_tree=None, in_seqs=None, in_meta=None, clock="4", out_tree=None, out_node=None):
    subprocess.run([
            "augur", "refine",
            "--tree", in_tree,
            "--alignment", in_seqs,
            "--metadata", in_meta,
            "--timetree",
            "--clock-filter-iqd", clock,
            "--output-tree", out_tree,
            "--output-node-data", out_node,
            "--stochastic-resolve"
        ], check=True)
```

```python
def augur_ancestral(in_tree=None, in_seqs=None, out_node=None, out_seqs=None):
    subprocess.run([
        "augur", "ancestral",
        "--tree", in_tree,
        "--alignment", in_seqs,
        "--output-node-data", out_node,
        "--output-sequences", out_seqs
    ], check=True)
```

```python
def augur_translate(in_tree=None, ancestral_seqs=None, ref_seq=None, genes=None, out_node=None, out_aln=None):
    cmd = [
        "augur", "translate",
        "--tree", in_tree,
        "--ancestral-sequences", ancestral_seqs,
        "--reference-sequence", ref_seq,
        "--output-node-data", out_node,
        "--alignment-output", out_aln
    ]
    if genes:
        cmd.append("--genes")
        cmd.extend(genes if isinstance(genes, list) else [genes])
    subprocess.run(cmd, check=True)
```

## Plots

### Average Distance Same Year

```python
def plot_temporal_grantham_distance(
    df_matrices: List[pd.DataFrame],
    meta_path_template: str,
    title: str,
    serotypes: Optional[List[int]] = None,
    figsize: Tuple[int, int] = (12, 7),
    palette: str = "husl",
    save_path: Optional[str] = None
) -> Tuple[pd.DataFrame, plt.Figure]:
    """
    Calcula y grafica la distancia de Grantham promedio a lo largo del tiempo para múltiples serotipos.
    
    Args:
        df_matrices: Lista de DataFrames con matrices de distancia (orden: DENV1, DENV2, DENV3, DENV4)
        meta_path_template: Ruta con placeholder `{sero}` (ej: "metadata_clean_col_{sero}.tsv")
        serotypes: Lista de serotipos a procesar. Por defecto: [1, 2, 3, 4]
        figsize: Tamaño de la figura
        palette: Paleta de colores para seaborn
        title: Título del gráfico
        save_path: Ruta para guardar la imagen (opcional)
        
    Returns:
        Tuple con (DataFrame procesado, objeto Figure de matplotlib)
    """
    if serotypes is None:
        serotypes = [1, 2, 3, 4]
        
    plot_data = []
    year_ranges = []

    for sero, matrix in zip(serotypes, df_matrices):
        meta_path = meta_path_template.format(sero)
        meta = pd.read_csv(meta_path, sep="\t")
        meta["date"] = pd.to_datetime(meta["date"], format="mixed", errors="coerce")
        meta = meta.dropna(subset=["date"]).copy()
        meta["year"] = meta["date"].dt.year

        # Mapeo rápido para evitar merges múltiples
        strain_to_year = dict(zip(meta["strain"], meta["year"]))

        # Formato largo
        df_long = matrix.stack().reset_index()
        df_long.columns = ["seq_a", "seq_b", "distance"]
        
        # Asignar años y filtrar pares válidos
        df_long["year_a"] = df_long["seq_a"].map(strain_to_year)
        df_long["year_b"] = df_long["seq_b"].map(strain_to_year)
        df_long = df_long.dropna(subset=["year_a", "year_b"])
        df_long = df_long[df_long["seq_a"] != df_long["seq_b"]].copy()
        
        # Año promedio del par y agregación anual
        df_long["year"] = ((df_long["year_a"] + df_long["year_b"]) / 2).astype(int)
        avg = df_long.groupby("year")["distance"].mean()
        
        if avg.empty:
            continue
            
        plot_data.append({"sero": sero, "avg": avg})
        year_ranges.append((avg.index.min(), avg.index.max()))

    if not plot_data:
        raise ValueError("No se encontraron datos válidos en ninguna serie temporal.")

    # Rango global y reindexación honesta (NaN para años sin datos)
    global_min = min(r[0] for r in year_ranges)
    global_max = max(r[1] for r in year_ranges)
    global_years = range(global_min, global_max + 1)

    final_data = []
    for item in plot_data:
        avg_smooth = item["avg"].reindex(global_years, fill_value=np.nan)
        df_temp = avg_smooth.reset_index().rename(columns={"index": "year", "distance": "avg_distance"})
        df_temp["serotype"] = f"DENV{item['sero']}"
        final_data.append(df_temp)

    plot_df = pd.concat(final_data, ignore_index=True)

    # Plot
    fig, ax = plt.subplots(figsize=figsize)
    sns.lineplot(
        data=plot_df, x="year", y="avg_distance", hue="serotype",
        marker="o", linewidth=2, palette=palette, ax=ax
    )
    ax.set_xlabel("Año", fontsize=12, fontweight="bold")
    ax.set_ylabel("Distancia Grantham promedio", fontsize=12, fontweight="bold")
    ax.set_title(title, fontsize=14, fontweight="bold", pad=20)
    ax.set_xticks(list(global_years))
    ax.set_xticklabels(list(global_years), rotation=45, ha="right")
    ax.legend(title="Serotipo", title_fontsize=11, fontsize=10)
    ax.grid(axis="y", alpha=0.3)
    sns.despine(ax=ax, top=True, right=True)
    plt.tight_layout()

    if save_path:
        plt.savefig(save_path, dpi=300, bbox_inches="tight")
    plt.show()
    plt.close()

    return plot_df, fig
```

### Average Distance Last Year

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from typing import List, Optional, Tuple

def plot_grantham_distance_to_latest(
    df_matrices: List[pd.DataFrame],
    meta_path_template: str,
    serotypes: Optional[List[int]] = None,
    figsize: Tuple[int, int] = (12, 7),
    palette: str = "husl",
    title: str = "Cambio evolutivo en el tiempo respecto a la variante más reciente",
    save_path: Optional[str] = None
) -> Tuple[pd.DataFrame, plt.Figure]:
    """
    Calcula y grafica la distancia de Grantham promedio de cada año frente 
    a la variante más reciente de cada serotipo.
    """
    if serotypes is None:
        serotypes = [1, 2, 3, 4]
        
    plot_data = []
    year_ranges = []

    for sero, matrix in zip(serotypes, df_matrices):
        meta_path = meta_path_template.format(sero)
        meta = pd.read_csv(meta_path, sep="\t")
        meta["date"] = pd.to_datetime(meta["date"], format="mixed", errors="coerce")
        meta = meta.dropna(subset=["date"]).copy()
        meta["year"] = meta["date"].dt.year
        
        if meta.empty:
            continue
            
        max_year = int(meta["year"].max())
        strain_to_year = dict(zip(meta["strain"], meta["year"]))

        # Formato largo
        df_long = matrix.stack().reset_index()
        df_long.columns = ["seq_a", "seq_b", "distance"]
        
        # Asignar años vía map (más rápido que merge)
        df_long["year_a"] = df_long["seq_a"].map(strain_to_year)
        df_long["year_b"] = df_long["seq_b"].map(strain_to_year)
        df_long = df_long.dropna(subset=["year_a", "year_b"]).copy()
        df_long["year_a"] = df_long["year_a"].astype(int)
        df_long["year_b"] = df_long["year_b"].astype(int)
        
        # Filtrar solo pares donde seq_b pertenece al año más reciente
        df_long = df_long[df_long["year_b"] == max_year].copy()
        df_long["year"] = df_long["year_a"]
        
        # Eliminar auto-comparaciones
        df_long = df_long[df_long["seq_a"] != df_long["seq_b"]]
        
        if df_long.empty:
            continue
            
        avg = df_long.groupby("year")["distance"].mean()
        plot_data.append({"sero": sero, "avg": avg})
        year_ranges.append((int(avg.index.min()), int(avg.index.max())))

    if not plot_data:
        raise ValueError("No se encontraron datos válidos en ninguna serie temporal.")

    # Rango global y reindexación honesta (NaN para años sin datos)
    global_min = min(r[0] for r in year_ranges)
    global_max = max(r[1] for r in year_ranges)
    global_years = range(global_min, global_max + 1)

    final_data = []
    for item in plot_data:
        avg_smooth = item["avg"].reindex(global_years, fill_value=np.nan)
        df_temp = avg_smooth.reset_index().rename(columns={"index": "year", "distance": "avg_distance"})
        df_temp["serotype"] = f"DENV{item['sero']}"
        final_data.append(df_temp)

    plot_df = pd.concat(final_data, ignore_index=True)

    # Plot
    fig, ax = plt.subplots(figsize=figsize)
    sns.lineplot(
        data=plot_df, x="year", y="avg_distance", hue="serotype",
        marker="o", linewidth=2, palette=palette, ax=ax
    )
    ax.set_xlabel("Año", fontsize=12, fontweight="bold")
    ax.set_ylabel("Distancia Grantham promedio", fontsize=12, fontweight="bold")
    ax.set_title(title, fontsize=14, fontweight="bold", pad=20)
    ax.set_xticks(list(global_years))
    ax.set_xticklabels(list(global_years), rotation=45, ha="right")
    ax.legend(title="Serotipo", title_fontsize=11, fontsize=10)
    ax.grid(axis="y", alpha=0.3)
    sns.despine(ax=ax, top=True, right=True)
    plt.tight_layout()

    if save_path:
        plt.savefig(save_path, dpi=300, bbox_inches="tight")
    plt.show()
    plt.close()

    return plot_df, fig
```

### Average Distance Against Reference

```python
import matplotlib.pyplot as plt
import seaborn as sns
from typing import Tuple, Optional

def plot_grantham_distance_temporal(
    df_distances: pd.DataFrame,
    figsize: Tuple[int, int] = (12, 7),
    palette: str = "husl",
    title: str = "Divergencia temporal respecto a la Referencia (gen E) - DENV1–4",
    save_path: Optional[str] = None
) -> Tuple[pd.DataFrame, plt.Figure]:
    """
    Agrega distancias por año/serotipo, reindexa al rango global y genera la línea temporal.
    
    Args:
        df_distances: DataFrame con columnas ['serotype', 'strain', 'distance', 'year']
        
    Returns:
        Tuple con (DataFrame agregado para plotting, objeto Figure)
    """
    if df_distances.empty:
        raise ValueError("El DataFrame de distancias está vacío.")
        
    # 1. Agregar promedio anual por serotipo
    yearly_avg = df_distances.groupby(["serotype", "year"])["distance"].mean().reset_index()
    
    # 2. Determinar rango global de años
    global_min = int(yearly_avg["year"].min())
    global_max = int(yearly_avg["year"].max())
    global_years = range(global_min, global_max + 1)
    
    # 3. Reindexar honestamente (NaN para años sin datos)
    plot_data = []
    for sero in sorted(df_distances["serotype"].unique()):
        sero_series = yearly_avg[yearly_avg["serotype"] == sero].set_index("year")["distance"]
        sero_smooth = sero_series.reindex(global_years, fill_value=np.nan).reset_index()
        sero_smooth.columns = ["year", "avg_dist_ref"]
        sero_smooth["serotype"] = f"DENV{sero}"
        plot_data.append(sero_smooth)
        
    plot_df = pd.concat(plot_data, ignore_index=True)
    
    # 4. Graficar
    sns.set_theme(style="whitegrid", context="notebook", palette=palette)
    fig, ax = plt.subplots(figsize=figsize)
    sns.lineplot(
        data=plot_df, x="year", y="avg_dist_ref", hue="serotype",
        marker="o", linewidth=2, palette=palette, ax=ax
    )
    ax.set_xlabel("Año", fontsize=12, fontweight="bold")
    ax.set_ylabel("Distancia Grantham vs Referencia", fontsize=12, fontweight="bold")
    ax.set_title(title, fontsize=14, fontweight="bold", pad=20)
    ax.set_xticks(list(global_years))
    ax.set_xticklabels(list(global_years), rotation=45, ha="right")
    ax.legend(title="Serotipo", title_fontsize=11, fontsize=10)
    ax.grid(axis="y", alpha=0.3)
    sns.despine(ax=ax, top=True, right=True)
    plt.tight_layout()
    
    if save_path:
        plt.savefig(save_path, dpi=300, bbox_inches="tight")
    plt.show()
    plt.close()
    
    return plot_df, fig
```

### Plot closest experimental match

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from typing import List, Optional, Tuple

def plot_temporal_grantham_distance_colombia(
    df_matrices: List[pd.DataFrame],
    meta_path_template: str,
    title: str,
    serotypes: Optional[List[int]] = None,
    figsize: Tuple[int, int] = (12, 7),
    palette: str = "husl",
    save_path: Optional[str] = None
) -> Tuple[pd.DataFrame, plt.Figure]:
    """
    Calcula y grafica la distancia de Grantham promedio a lo largo del tiempo,
    usando SOLAMENTE los años de las secuencias colombianas para el eje temporal.
    
    Args:
        df_matrices: Lista de DataFrames con matrices de distancia cruzada 
                     (index=Colombia, columns=Experimental)
        meta_path_template: Ruta con placeholder `{}` a metadatos fusionados 
                           (debe contener años para ambos conjuntos de IDs)
        title: Título del gráfico
        serotypes: Lista de serotipos. Por defecto: [1, 2, 3, 4]
        figsize: Tamaño de la figura
        palette: Paleta de colores para seaborn
        save_path: Ruta para guardar la imagen (opcional)
        
    Returns:
        Tuple con (DataFrame procesado, objeto Figure de matplotlib)
    """
    if serotypes is None:
        serotypes = [1, 2, 3, 4]
        
    plot_data = []
    year_ranges = []

    for sero, matrix in zip(serotypes, df_matrices):
        meta_path = meta_path_template.format(sero)
        meta = pd.read_csv(meta_path, sep="\t")
        meta["date"] = pd.to_datetime(meta["date"], format="mixed", errors="coerce")
        meta = meta.dropna(subset=["date"]).copy()
        meta["year"] = meta["date"].dt.year

        # Mapeo rápido para evitar merges múltiples
        strain_to_year = dict(zip(meta["strain"], meta["year"]))

        # Formato largo: matrix tiene index=Colombia, columns=Experimental
        df_long = matrix.stack().reset_index()
        df_long.columns = ["seq_col", "seq_exp", "distance"]
        
        # Asignar años: seq_col = colombiana, seq_exp = experimental
        df_long["year_col"] = df_long["seq_col"].map(strain_to_year)
        df_long["year_exp"] = df_long["seq_exp"].map(strain_to_year)
        
        # Filtrar: mantener solo pares con año colombiano válido
        df_long = df_long.dropna(subset=["year_col"]).copy()
        df_long = df_long[df_long["seq_col"] != df_long["seq_exp"]].copy()
        
        if df_long.empty:
            continue
            
        # 🔑 CAMBIO CLAVE: Usar SOLO el año colombiano para el eje temporal
        df_long["year"] = df_long["year_col"].astype(int)
        
        # Agregación anual
        avg = df_long.groupby("year")["distance"].mean()
        
        if avg.empty:
            continue
            
        plot_data.append({"sero": sero, "avg": avg})
        year_ranges.append((int(avg.index.min()), int(avg.index.max())))

    if not plot_data:
        raise ValueError("No se encontraron datos válidos para ningún serotipo.")

    # Rango global y reindexación honesta (NaN para años sin datos)
    global_min = min(r[0] for r in year_ranges)
    global_max = max(r[1] for r in year_ranges)
    global_years = range(global_min, global_max + 1)

    final_data = []
    for item in plot_data:
        avg_smooth = item["avg"].reindex(global_years, fill_value=np.nan)
        df_temp = avg_smooth.reset_index().rename(columns={"index": "year", "distance": "avg_distance"})
        df_temp["serotype"] = f"DENV{item['sero']}"
        final_data.append(df_temp)

    plot_df = pd.concat(final_data, ignore_index=True)

    # Visualización
    sns.set_theme(style="whitegrid", context="notebook", palette=palette)
    fig, ax = plt.subplots(figsize=figsize)
    sns.lineplot(
        data=plot_df, x="year", y="avg_distance", hue="serotype",
        marker="o", linewidth=2, palette=palette, ax=ax
    )
    ax.set_xlabel("Año", fontsize=12, fontweight="bold")
    ax.set_ylabel("Distancia Grantham promedio (EDIII)", fontsize=12, fontweight="bold")
    ax.set_title(title, fontsize=14, fontweight="bold", pad=20)
    ax.set_xticks(list(global_years))
    ax.set_xticklabels(list(global_years), rotation=45, ha="right")
    ax.legend(title="Serotipo", title_fontsize=11, fontsize=10)
    ax.grid(axis="y", alpha=0.3)
    sns.despine(ax=ax, top=True, right=True)
    plt.tight_layout()

    if save_path:
        plt.savefig(save_path, dpi=300, bbox_inches="tight")
    plt.show()
    plt.close()

    return plot_df, fig
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from typing import List, Optional, Tuple

def plot_closest_experimental_matches(
    df_matrices: List[pd.DataFrame],
    serotypes: Optional[List[int]] = None,
    top_n: int = 4,
    figsize: Tuple[int, int] = (14, 10),
    palette: str = "husl",
    title: str = "Distribución de secuencias experimentales con menor distancia",
    save_path: Optional[str] = None
) -> Tuple[pd.DataFrame, plt.Figure]:
    """
    Para cada secuencia colombiana, identifica la secuencia experimental más cercana
    (mínima distancia de Grantham) y grafica la frecuencia de las top-N referencias.
    
    Args:
        df_matrices: Lista de DataFrames rectangulares (index=Colombia, cols=Experimental)
        serotypes: Lista de serotipos. Por defecto: [1, 2, 3, 4]
        top_n: Número de secuencias experimentales "top" a mostrar por gráfico
        figsize: Tamaño de la figura
        palette: Paleta de colores para seaborn
        title: Título principal del gráfico
        save_path: Ruta para guardar la imagen (opcional)
        
    Returns:
        Tuple con (DataFrame con resultados detallados, objeto Figure)
    """
    if serotypes is None:
        serotypes = [1, 2, 3, 4]
    
    sns_colors = sns.color_palette(palette, n_colors=len(serotypes))
    results = []  # Para almacenar resultados detallados
    
    fig, axes = plt.subplots(2, 2, figsize=figsize)
    axes = axes.flatten()
    
    for idx, (sero, matrix) in enumerate(zip(serotypes, df_matrices)):
        ax = axes[idx]
        
        # 1. Encontrar el índice de la columna (seq_exp) con valor mínimo por fila (seq_col)
        # idxmin(axis=1) ignora NaN por defecto
        closest_exp = matrix.idxmin(axis=1)
        min_dist = matrix.min(axis=1)
        
        # 2. Filtrar filas donde no hubo coincidencia válida (todo NaN)
        valid_mask = ~min_dist.isna()
        closest_exp = closest_exp[valid_mask]
        min_dist = min_dist[valid_mask]
        
        if closest_exp.empty:
            ax.text(0.5, 0.5, "Sin datos válidos", ha='center', va='center', transform=ax.transAxes)
            ax.set_title(f'DENV{sero}')
            continue
        
        # 3. Guardar resultados para análisis posterior
        df_res = pd.DataFrame({
            "serotype": sero,
            "seq_col": closest_exp.index,
            "closest_exp": closest_exp.values,
            "min_distance": min_dist.values
        })
        results.append(df_res)
        
        # 4. Contar frecuencias de las secuencias experimentales más cercanas
        exp_counts = closest_exp.value_counts().nlargest(top_n)
        
        if exp_counts.empty:
            ax.text(0.5, 0.5, "Sin coincidencias", ha='center', va='center', transform=ax.transAxes)
            ax.set_title(f'DENV{sero}')
            continue
        
        # 5. Plotear
        bars = ax.bar(exp_counts.index, exp_counts.values, color=sns_colors[idx], edgecolor='black')
        ax.tick_params(axis='x', rotation=45, labelsize=9)
        ax.set_title(f'DENV{sero}\n(n={len(closest_exp)} secuencias colombianas)', fontsize=11, fontweight="bold")
        ax.set_xlabel("Secuencia experimental de referencia")
        ax.set_ylabel("Número de cepas colombianas más cercanas")
        ax.grid(axis='y', alpha=0.3, linestyle='--')
        
        # Añadir etiquetas de valor en las barras
        ax.bar_label(bars, padding=2, fontsize=8)
    
    # Título general y ajuste final
    fig.suptitle(title, fontsize=14, fontweight="bold", y=1.02)
    plt.tight_layout()
    
    if save_path:
        plt.savefig(save_path, dpi=300, bbox_inches="tight")
    plt.show()
    plt.close()
    
    # Concatenar todos los resultados en un solo DataFrame
    df_results = pd.concat(results, ignore_index=True) if results else pd.DataFrame()
    
    return df_results, fig
```

## Crop Aligments Gen E

```python
for i in range(4):
    print(f"Crop DENV{i+1}")
    serotype = i + 1
    db = [db_denv1, db_denv2, db_denv3, db_denv4][i]
    e_start, e_end, strand = get_feature_coords(db, "envelope protein E")
    cropped_records = []
    
    for j, record in enumerate(SeqIO.parse(ALN_SEQS_EXP.format(serotype), "fasta")):
        
        if j == 0:
            continue
            
        if len(record.seq) < e_end:
            continue
        
        cropped_seq = record.seq[e_start-1:e_end]
        
        record.seq = cropped_seq

        new_id = record.id.split("|")[0]
        
        record.id = new_id
        record.name = new_id
        record.description = new_id
        
        cropped_records.append(record)
    
    SeqIO.write(
        cropped_records,
        CROP_SEQS_EXP.format(serotype),
        "fasta"
    )
```

## Create Metadata

```python
for i in range(4):
    sero_num = i + 1
    crop_fasta = CROP_SEQS_EXP.format(sero_num)
    meta_path = META_EXP.format(sero_num)

    print(f"Step 2: Processing DENV{sero_num} Metadata")

    meta_rows = []

    for record in SeqIO.parse(crop_fasta, "fasta"):
        parts = record.id.split("/")
        
        if len(parts) < 3:
            print(f"  Warning: Skipping {record.id}, not enough fields.")
            continue

        strain_name = record.id
        country = parts[1]
        year = parts[2]

        date_str = f"{year}-01-01" if year.isdigit() and len(year) == 4 else None

        meta_rows.append({
            "strain": strain_name,
            "date": date_str,
            "country": country,
            "genotype": parts[0],
            "accession": parts[3] if len(parts) > 3 else ".",
            "extra": parts[4] if len(parts) > 4 else "."
        })

    meta_df = pd.DataFrame(meta_rows)
    
    required_cols = ["strain", "date", "country"]
    missing_cols = [c for c in required_cols if c not in meta_df.columns]
    if missing_cols:
        raise ValueError(f"DENV{sero_num}: Missing required metadata columns: {missing_cols}")

    before_filter = len(meta_df)
    meta_df = meta_df.dropna(subset=["date"])
    print(f"  Meta {len(meta_df)}/{before_filter} strains retained with valid dates.")

    meta_df.to_csv(meta_path, sep="\t", index=False)
    print(f"  Saved: {meta_path}\n")
```

# Merged Data

```python
def merge_fastas(f1, f2, out):
    Path(out).parent.mkdir(parents=True, exist_ok=True)
    seen_ids = set()
    with open(out, "w") as out_f:
        for src, suffix in [(f1, "_COL"), (f2, "_EXP")]:
            with open(src, "r") as in_f:
                current_id = None
                for line in in_f:
                    if line.startswith(">"):
                        current_id = line[1:].strip().split()[0]
                        if current_id in seen_ids:
                            current_id = f"{current_id}{suffix}"
                            line = f">{current_id}\n"
                        seen_ids.add(current_id)
                    out_f.write(line)
    print(f"FASTA → {out}")

def merge_tsvs(f1, f2, out):
    Path(out).parent.mkdir(parents=True, exist_ok=True)
    header = None
    with open(out, "w") as out_f:
        for src in (f1, f2):
            with open(src, "r") as in_f:
                for line in in_f:
                    stripped = line.strip()
                    # Preserve comments & empty lines
                    if stripped.startswith("#") or stripped == "":
                        out_f.write(line)
                        continue
                    # First non-comment line = header
                    if header is None:
                        header = stripped
                        out_f.write(line)
                    # Skip duplicate headers, write data
                    elif stripped != header:
                        out_f.write(line)
    print(f"TSV  → {out}")

def merge_gffs(f1, f2, out):
    """Merges GFF3 files, ensures ##gff-version appears only once at top."""
    Path(out).parent.mkdir(parents=True, exist_ok=True)
    version_written = False
    with open(out, "w") as out_f:
        for src in (f1, f2):
            with open(src, "r") as in_f:
                for line in in_f:
                    stripped = line.strip()
                    if stripped.startswith("##gff-version"):
                        if not version_written:
                            out_f.write(line)
                            version_written = True
                        continue
                    out_f.write(line)
    print(f"GFF  → {out}")


SEROTYPES = [1, 2, 3, 4]

for sero in SEROTYPES:
    print(f"\n🔹 Processing DENV-{sero}...")
    merge_fastas(
        CROP_SEQS_COL.format(sero),
        CROP_SEQS_EXP.format(sero),
        CROP_SEQS_MERGE.format(sero)
    )
    merge_tsvs(
        META_CLEAN_COL.format(sero),
        META_EXP.format(sero),
        META_MERGE.format(sero)
    )
    merge_gffs(
        TMP_GFF_COL.format(sero),
        TMP_GFF_EXP.format(sero),
        TMP_GFF_MERGE.format(sero)
    )

print("All merges complete.")
```

```python
warnings.filterwarnings("ignore", category=SyntaxWarning)

fig, axes = plt.subplots(2, 2, figsize=(16, 14))
axes = axes.flatten()

for idx, s in enumerate(range(4)):
    sero = s + 1
    ax = axes[idx]
    
    ll = bt.loadNewick(TIME_TREE_MERGE.format(sero))
    df = pd.read_csv(META_MERGE.format(sero), sep='\t')
    df['country'] = df['country'].fillna('').str.strip().str.lower()
    
    origin_map = {}
    for _, row in df.iterrows():
        strain = row['strain'].strip()
        origin = 'COL' if 'colombia' in row['country'] else 'EXP'
        origin_map[strain] = origin
        for token in strain.replace('/', ' ').replace('-', ' ').replace('_', ' ').split():
            if len(token) > 2:
                origin_map[token] = origin
                
    leaf_colors = {}
    for k in ll.Objects:
        if k.branchType == 'leaf':
            name = k.name
            match = origin_map.get(name)
            if not match:
                for t in name.replace('/', ' ').replace('-', ' ').replace('_', ' ').split():
                    if t in origin_map:
                        match = origin_map[t]
                        break
            leaf_colors[name] = "#96c7ea" if match == 'COL' else ("#ffc99b" if match == 'EXP' else "#888888")
            
    def c_func(k):
        if k.branchType == 'leaf':
            return leaf_colors.get(k.name, "#888888")
        has_col = has_exp = False
        for d in k.leaves:
            d_name = d.name if hasattr(d, 'name') else d
            c = leaf_colors.get(d_name, "#888888")
            if c == "#96c7ea": has_col = True
            if c == "#ffc99b": has_exp = True
            if has_col and has_exp: return "#888888"
        return "#96c7ea" if has_col else ("#ffc99b" if has_exp else "#888888")
        
    L = sum(1 for k in ll.Objects if k.branchType == 'leaf')
    b_func = lambda k: 2 + 10.0 * len(k.leaves) / L if k.branchType == 'node' else 2
    
    ll.plotTree(ax, width=b_func, colour=c_func, zorder=20000)
    ll.plotPoints(ax, size=30, colour=c_func, zorder=20002)
    
    for loc in ['left', 'right', 'top']: ax.spines[loc].set_visible(False)
    ax.grid(axis='x', ls='-', color='grey', alpha=0.3)
    ax.tick_params(axis='y', size=0)
    ax.tick_params(axis='x', labelsize=16)
    ax.set_xlabel('Time / Divergence', fontsize=16)
    ax.set_yticklabels([])
    ax.set_title(f"DENV {sero}", fontsize=16, fontweight='bold')

legend_elements = [
    Line2D([0], [0], marker='o', color='w', label='Colombia Sequences', markerfacecolor='#96c7ea', markersize=12),
    Line2D([0], [0], marker='o', color='w', label='Antigenic Map sequences', markerfacecolor='#ffc99b', markersize=12)
]
fig.legend(handles=legend_elements, loc='lower center', bbox_to_anchor=(0.5, 0.02),
           fontsize=14, frameon=True, edgecolor='black', ncol=2)

plt.tight_layout()
plt.savefig("../figures/DENV_all_serotypes_col_exp.jpeg", dpi=300, bbox_inches='tight')
plt.show()
plt.close()
```

```python
results = {}

for s in range(4):
    sero = s + 1

    records_col = [r for r in SeqIO.parse(AA_SEQS_COL_E.format(sero), "fasta") if not r.id.startswith("NODE")]
    ids_col = [r.id for r in records_col]
    seqs_col = [str(r.seq) for r in records_col]

    records_exp = [r for r in SeqIO.parse(AA_SEQS_EXP_E.format(sero), "fasta") if not r.id.startswith("NODE")]
    ids_exp = [r.id for r in records_exp]
    seqs_exp = [str(r.seq) for r in records_exp]

    n, m = len(seqs_col), len(seqs_exp)
    matrix = np.zeros((n, m))
    for i in range(n):
        for j in range(m):
            dist,sites = grantham_distance(seqs_col[i], seqs_exp[j])
            matrix[i, j] = dist/sites

    matrix_scaled = StandardScaler().fit_transform(matrix)

    perplexity_val = min(30, max(5, len(ids_col) - 1))
    tsne = TSNE(n_components=2, perplexity=perplexity_val, learning_rate='auto', 
                max_iter=1000, random_state=42, metric='euclidean', init='pca')
    tsne_coords = tsne.fit_transform(matrix_scaled)

    df_tsne = pd.DataFrame(tsne_coords, columns=["tSNE1", "tSNE2"], index=ids_col)
    df_tsne = df_tsne.reset_index().rename(columns={"index": "strain"})
    df_tsne["lineage"] = np.nan

    meta_col = pd.read_csv(META_CLEAN_COL.format(sero), sep="\t")
    lineage_col = next((c for c in meta_col.columns if 'lineage' in c.lower()), None)

    df_merge = meta_col[["strain", lineage_col]].drop_duplicates(subset=["strain"]).copy()
    df_merge.rename(columns={lineage_col: "lineage"}, inplace=True)
    if "lineage" in df_tsne.columns:
        df_tsne.drop(columns=["lineage"], inplace=True)
    df_tsne = df_tsne.merge(df_merge, on="strain", how="left")

    results[sero] = df_tsne.copy()

fig, axes = plt.subplots(2, 2, figsize=(16, 14))
axes = axes.flatten()

for idx, sero in enumerate([1, 2, 3, 4]):
    ax = axes[idx]
    df = results[sero]
    df_plot = df.dropna(subset=["lineage"])

    sns.scatterplot(
        data=df_plot, x="tSNE1", y="tSNE2",
        hue="lineage", palette="tab10", s=80, alpha=0.85, ax=ax
    )
    ax.set_title(f"DENV {sero}\n(n={len(df)} secuencias)", fontsize=13, fontweight="bold")
    ax.set_xlabel("t-SNE 1", fontsize=11)
    ax.set_ylabel("t-SNE 2", fontsize=11)
    ax.legend(title="Lineage", loc="best", fontsize=9, title_fontsize=10)

plt.tight_layout()
plt.savefig("tsne_grid_4serotypes_lineage.png", dpi=300, bbox_inches="tight")
plt.show()
plt.close()
```

# Antigenic Space Projection

```python
import numpy as np
import pandas as pd
from Bio import SeqIO

# Global cache to store calculated Grantham distances
grantham_cache = {}

def get_grantham_dist(seq1, seq2):
    # Ensure symmetry to halve cache size (dist(A,B) == dist(B,A))
    if seq1 > seq2:
        seq1, seq2 = seq2, seq1
    key = (seq1, seq2)
    if key not in grantham_cache:
        grantham_cache[key] = grantham_distance(seq1, seq2)
    return grantham_cache[key]

all_exp_records = []
for sero in [1, 2, 3, 4]:
    records = [r for r in SeqIO.parse(AA_SEQS_EXP_E.format(sero), "fasta") 
               if not r.id.startswith("NODE")]
    all_exp_records.extend(records)

all_exp_ids = [r.id for r in all_exp_records]
all_exp_seqs = [str(r.seq[296:397]) for r in all_exp_records]  # EDIII region
n_all_exp = len(all_exp_seqs)

colombian_grantham_matrices = []

for sero in [1, 2, 3, 4]:
    print(f"Computing Grantham distances for DENV{sero}...")
    
    col_records = [r for r in SeqIO.parse(AA_SEQS_COL_E.format(sero), "fasta") 
                   if not r.id.startswith("NODE")]
    col_ids = [r.id for r in col_records]
    col_seqs = [str(r.seq[296:397]) for r in col_records]  # EDIII region
    
    n_col = len(col_seqs)
    matrix = np.zeros((n_col, n_all_exp))
    
    for i in range(n_col):
        for j in range(n_all_exp):
            dist, sites = get_grantham_dist(col_seqs[i], all_exp_seqs[j])
            matrix[i, j] = dist if sites > 0 else np.nan
    
    df_matrix = pd.DataFrame(matrix, index=col_ids, columns=all_exp_ids)
    colombian_grantham_matrices.append(df_matrix)
    
    print(f"  Matrix shape: {df_matrix.shape}")
    print(f"  Colombian strains: {len(col_ids)}")
    print(f"  Unified Experimental references: {len(all_exp_ids)}")
```

```python
for i, sero in enumerate([1, 2, 3, 4]):
    filename = f"../output/grantham_distances_DENV{sero}.csv"
    colombian_grantham_matrices[i].to_csv(filename, index=True, na_rep='NaN')
    
    print(f"Matriz de DENV{sero} exportada a: {filename}")
```

```python
mapa_ag = pd.read_csv(
    "../resources/experimental_data/3d-antigenic-map.txt",
    sep=r'\s+',
    header=None,
    names=['type', 'strainID', 'coord1', 'coord2', 'coord3']
)

mapa_ag["sero_num"] = mapa_ag["strainID"].str.extract(r'DEN(?:V)?(\d)')[0].astype(float)
mapa_ag["sero"] = "DENV" + mapa_ag["sero_num"].astype(int).astype(str)

ref_coords_by_sero = {}
reference_centroids = {}

for sero_num in [1, 2, 3, 4]:
    sero_data = mapa_ag[mapa_ag['sero_num'] == sero_num].copy()
    
    if len(sero_data) > 0:
        coords_df = sero_data.set_index('strainID')[['coord1', 'coord2', 'coord3']].copy()
        coords_df.columns = ['X', 'Y', 'Z']
        ref_coords_by_sero[sero_num] = coords_df
        
        ref_centroid = coords_df.mean().to_frame().T
        ref_centroid.index = [f'DENV{sero_num}_REF_Centroid']
        reference_centroids[sero_num] = ref_centroid
        
        print(f"DENV{sero_num}: {len(coords_df)} reference strains loaded")
        print(f"  Reference Centroid (X, Y, Z): ({ref_centroid['X'].values[0]:.4f}, {ref_centroid['Y'].values[0]:.4f}, {ref_centroid['Z'].values[0]:.4f})")
    else:
        print(f"DENV{sero_num}: No reference strains found")

print("\nReference antigenic coordinates and centroids loaded successfully")
```

```python
mapa_ag_indexed = mapa_ag.drop_duplicates(subset='strainID').set_index('strainID')
ref_coords = mapa_ag_indexed.reindex(all_exp_ids)[['coord1', 'coord2', 'coord3']].fillna(0).values

projected_coordinates = {}
validation_results = {}

for sero in [1, 2, 3, 4]:
    print(f"DENV{sero}:")
    
    grantham_matrix = colombian_grantham_matrices[sero - 1]
    colombian_strains = grantham_matrix.index.tolist()
    
    D = grantham_matrix.values.astype(float)
    
    for i in range(D.shape[0]):
        nan_mask = np.isnan(D[i, :])
        if nan_mask.all():
            D[i, :] = 1.0
        else:
            D[i, nan_mask] = np.nanmax(D[i, :])
            
    projected = np.zeros((len(colombian_strains), 3))
    
    for i in range(len(colombian_strains)):
        d_i = D[i, :]
        weights = 1.0 / (d_i + 1e-6)
        weights = weights / weights.sum()
        a_init = (weights[:, np.newaxis] * ref_coords).sum(axis=0)
        
        def residuals(a):
            euclidean_dists = np.linalg.norm(ref_coords - a, axis=1)
            max_d = np.max(d_i)
            d_norm = d_i / max_d if max_d > 0 else d_i
            return weights * (euclidean_dists - d_norm)
        
        try:
            result = least_squares(residuals, a_init, method='lm', max_nfev=100,
                                 ftol=1e-8, xtol=1e-8, gtol=1e-8)
            projected[i] = result.x
        except Exception as e:
            print(f"    Warning: Optimization failed for {colombian_strains[i]}. Using initial guess.")
            projected[i] = a_init
    
    df_proj = pd.DataFrame(projected, index=colombian_strains, columns=['X', 'Y', 'Z'])
    projected_coordinates[sero] = df_proj
```
