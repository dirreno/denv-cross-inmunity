```python
def hamming_distance(s1, s2):
    if len(s1) != len(s2):
        raise ValueError("Strings must be of the same length")
    
    return sum(el1 != el2 for el1, el2 in zip(s1, s2))
```

```python
folder_path = "../output/db"
os.makedirs(folder_path, exist_ok=True)

def load_or_create_db(gff_path, db_name):
    if not os.path.exists(db_name):
        print(f"Creating {db_name}...")
        gffutils.create_db(
            gff_path,
            dbfn=db_name,
            force=True,
            keep_order=True,
            merge_strategy="merge",
            sort_attribute_values=True
        )
    else:
        print(f"{db_name} already exists. Loading...")

    return gffutils.FeatureDB(db_name)


db_denv1 = load_or_create_db("../resources/reference_data/metadata/meta_ref_DENV1.gff3", "../output/db/denv1.db")
db_denv2 = load_or_create_db("../resources/reference_data/metadata/meta_ref_DENV2.gff3", "../output/db/denv2.db")
db_denv3 = load_or_create_db("../resources/reference_data/metadata/meta_ref_DENV3.gff3", "../output/db/denv3.db")
db_denv4 = load_or_create_db("../resources/reference_data/metadata/meta_ref_DENV4.gff3", "../output/db/denv4.db")
```

### Get Feature Coords

```python
def get_feature_coords(db, feature_name):
    for f in db.features_of_type("mature_protein_region_of_CDS"):
        product_name = f.attributes.get("product", [""])[0]
        if feature_name in product_name:
            return f.start, f.end, f.strand
    return None
```

### Compute Grantham Matrices

```python
def compute_grantham_matrices(
    fasta_path_template: str,
    grantham_func: Callable[[str, str], Tuple[float, int]],
    serotypes: Optional[List[int]] = None,
    ediii_start: int = 297,
    ediii_end: int = 397
) -> List[pd.DataFrame]:
    
    if serotypes is None:
        serotypes = [1, 2, 3, 4]
        
    matrices = []
    
    for sero in serotypes:
        records = list(SeqIO.parse(fasta_path_template.format(sero), "fasta"))
            
        ids = [r.id for r in records]
        seqs = [str(r.seq) for r in records]
        n = len(seqs)
        
        # Pre-cortar región EDIII para mayor eficiencia
        ediii_seqs = [s[ediii_start:ediii_end] for s in seqs]
        
        # Usar NaN en lugar de -1 para compatibilidad con pandas/numpy
        matrix = np.full((n, n), np.nan)
        
        for i in range(n):
            for j in range(i, n):
                dist, sites = grantham_func(ediii_seqs[i], ediii_seqs[j])
                value = dist / sites if sites > 0 else np.nan
                matrix[i, j] = value
                matrix[j, i] = value
                
        df_matrix = pd.DataFrame(matrix, index=ids, columns=ids)
        matrices.append(df_matrix)
        
    return matrices
```

```python
def compute_grantham_distances_vs_reference(
    fasta_path_template: str,
    ref_path_template: str,
    meta_path_template: str,
    grantham_func: Callable[[str, str], Tuple[float, int]],
    serotypes: Optional[List[int]] = None
) -> pd.DataFrame:
    
    if serotypes is None:
        serotypes = [1, 2, 3, 4]
        
    all_data = []
    
    for sero in serotypes:
        records = list(SeqIO.parse(fasta_path_template.format(sero), "fasta"))
            
        ids = [r.id.split('|')[0] if '|' in r.id else r.id for r in records]
        seqs = [str(r.seq) for r in records]

        ref_records = list(SeqIO.parse(ref_path_template.format(sero), "fasta"))
        ref_seq = str(ref_records[0].seq)
        
        distances = []
        for seq in seqs:
            dist, sites = grantham_func(seq, ref_seq)
            distances.append(dist / sites if sites > 0 else np.nan)
            
        df_dist = pd.DataFrame({"strain": ids, "distance": distances}).dropna(subset=["distance"])
            
        meta = pd.read_csv(meta_path_template.format(sero), sep="\t")
        meta["date"] = pd.to_datetime(meta["date"], format="mixed", errors="coerce")
        meta = meta.dropna(subset=["date"]).copy()
        meta["year"] = meta["date"].dt.year
            
        df_combined = df_dist.merge(meta[["strain", "year"]], on="strain", how="inner")
            
        df_combined["serotype"] = sero
        all_data.append(df_combined[["serotype", "strain", "distance", "year"]])
        
    return pd.concat(all_data, ignore_index=True)
```

```python
def compute_grantham_cross_matrices_for_plotter(
    fasta_col_template: str,
    fasta_exp_template: str,
    meta_col_template: str,
    meta_exp_template: str,
    grantham_func: Callable[[str, str], Tuple[float, int]],
    serotypes: Optional[List[int]] = None,
    region_slice: Optional[Tuple[int, int]] = (297, 397),
    exclude_prefix: str = "NODE",
    output_meta_dir: str = "merged_metadata"
) -> Tuple[List[pd.DataFrame], str]:
    
    if serotypes is None:
        serotypes = [1, 2, 3, 4]
        
    os.makedirs(output_meta_dir, exist_ok=True)
    matrices = []
    
    for sero in serotypes:
        records_col = [r for r in SeqIO.parse(fasta_col_template.format(sero), "fasta") 
                      if not r.id.startswith(exclude_prefix)]
        records_exp = [r for r in SeqIO.parse(fasta_exp_template.format(sero), "fasta") 
                      if not r.id.startswith(exclude_prefix)]
            
        ids_col = [r.id for r in records_col]
        seqs_col = [str(r.seq) for r in records_col]
        ids_exp = [r.id for r in records_exp]
        seqs_exp = [str(r.seq) for r in records_exp]
        
        if region_slice:
            start, end = region_slice
            seqs_col = [s[start:end] for s in seqs_col]
            seqs_exp = [s[start:end] for s in seqs_exp]
            
        n, m = len(seqs_col), len(seqs_exp)
        matrix = np.full((n, m), np.nan)
        for i in range(n):
            for j in range(m):
                dist, sites = grantham_func(seqs_col[i], seqs_exp[j])
                matrix[i, j] = dist / sites if sites > 0 else np.nan
                
        df_matrix = pd.DataFrame(matrix, index=ids_col, columns=ids_exp)
        matrices.append(df_matrix)
        
        meta_col = pd.read_csv(meta_col_template.format(sero), sep="\t")
        meta_exp = pd.read_csv(meta_exp_template.format(sero), sep="\t")

        for df in [meta_col, meta_exp]:
            if "strain" not in df.columns:
                df["strain"] = df.iloc[:, 0]
                
        meta_merged = pd.concat([meta_col, meta_exp], ignore_index=True).drop_duplicates(subset=["strain"])
        merged_path = f"{output_meta_dir}/meta_merged_{sero}.tsv"
        meta_merged.to_csv(merged_path, sep="\t", index=False)
        
    merged_template = f"{output_meta_dir}/meta_merged_{{}}.tsv"
    return matrices, merged_template
```

## Augur Functions

```python
def augur_align(in_seqs=None, ref_seq=None, out_aln=None, fill_gaps=False, nthreads="auto"):
    cmd = [
        "augur", "align",
        "--sequences", in_seqs,
        "--output", out_aln,
        "--nthreads", str(nthreads)
    ]
    if ref_seq:
        cmd.extend(["--reference-sequence", ref_seq])
    if fill_gaps:
        cmd.append("--fill-gaps")
    subprocess.run(cmd, check=True)
```

```python
def augur_tree(in_seqs=None, method="iqtree", sub_model="GTR+G", nthreads="auto", out_tree=None):
    subprocess.run([
            "augur", "tree",
            "--alignment", in_seqs,
            "--method", method,
            "--substitution-model", sub_model,
            "--nthreads", nthreads,
            "--output", out_tree
        ], check=True)
```

```python
def augur_refine(in_tree=None, in_seqs=None, in_meta=None, clock="4", out_tree=None, out_node=None):
    subprocess.run([
            "augur", "refine",
            "--tree", in_tree,
            "--alignment", in_seqs,
            "--metadata", in_meta,
            "--timetree",
            "--clock-filter-iqd", clock,
            "--output-tree", out_tree,
            "--output-node-data", out_node,
            "--stochastic-resolve"
        ], check=True)
```

```python
def augur_ancestral(in_tree=None, in_seqs=None, out_node=None, out_seqs=None):
    subprocess.run([
        "augur", "ancestral",
        "--tree", in_tree,
        "--alignment", in_seqs,
        "--output-node-data", out_node,
        "--output-sequences", out_seqs
    ], check=True)
```

```python
def augur_translate(in_tree=None, ancestral_seqs=None, ref_seq=None, genes=None, out_node=None, out_aln=None):
    cmd = [
        "augur", "translate",
        "--tree", in_tree,
        "--ancestral-sequences", ancestral_seqs,
        "--reference-sequence", ref_seq,
        "--output-node-data", out_node,
        "--alignment-output", out_aln
    ]
    if genes:
        cmd.append("--genes")
        cmd.extend(genes if isinstance(genes, list) else [genes])
    subprocess.run(cmd, check=True)
```

# Prepare Reference Data

```python
dbs = {1: db_denv1, 2: db_denv2, 3: db_denv3, 4: db_denv4}
feature_name = "envelope protein E"

for sero in range(1, 5):
    db = dbs[sero]
    ref_in = REF_SEQ.format(sero)
    crop_out = CROP_REF_SEQS.format(sero)
    aa_out = AA_REF_SEQ.format(sero)

    print(f"\nProcessing DENV{sero}...")

    coords = get_feature_coords(db, feature_name)
    if not coords:
        print(f"  Skipped: '{feature_name}' not found in DENV{sero} DB.")
        continue
    start, end, strand = coords
    print(f"  Found E region: {start}-{end}")

    if not os.path.exists(ref_in):
        raise FileNotFoundError(f"Reference FASTA not found: {ref_in}")
    ref_record = SeqIO.read(ref_in, "fasta")
    ref_seq = ref_record.seq

    cds_seq = ref_seq[start-1:end]

    with open(crop_out, "w") as f:
        f.write(f">{ref_record.id}_E_CDS\n")
        f.write(str(cds_seq) + "\n")
    print(f"  Saved nucleotide: {crop_out}")

    if len(cds_seq) % 3 != 0:
        print(f"  Warning: CDS length ({len(cds_seq)}) not multiple of 3. Padding with N...")
        cds_seq += Seq("N" * (3 - len(cds_seq) % 3))

    aa_seq = cds_seq.translate(table=1, to_stop=False)
    if str(aa_seq).endswith("*"):
        aa_seq = aa_seq[:-1]

    with open(aa_out, "w") as f:
        f.write(f">{ref_record.id}_E_protein\n")
        seq_str = str(aa_seq)
        for i in range(0, len(seq_str), 80):
            f.write(seq_str[i:i+80] + "\n")
    print(f"  Saved AA: {aa_out} ({len(aa_seq)} amino acids)")
```

# Merged Data

```python
def merge_fastas(f1, f2, out):
    Path(out).parent.mkdir(parents=True, exist_ok=True)
    seen_ids = set()
    with open(out, "w") as out_f:
        for src, suffix in [(f1, "_COL"), (f2, "_EXP")]:
            with open(src, "r") as in_f:
                current_id = None
                for line in in_f:
                    if line.startswith(">"):
                        current_id = line[1:].strip().split()[0]
                        if current_id in seen_ids:
                            current_id = f"{current_id}{suffix}"
                            line = f">{current_id}\n"
                        seen_ids.add(current_id)
                    out_f.write(line)
    print(f"FASTA → {out}")

def merge_tsvs(f1, f2, out):
    Path(out).parent.mkdir(parents=True, exist_ok=True)
    header = None
    with open(out, "w") as out_f:
        for src in (f1, f2):
            with open(src, "r") as in_f:
                for line in in_f:
                    stripped = line.strip()
                    # Preserve comments & empty lines
                    if stripped.startswith("#") or stripped == "":
                        out_f.write(line)
                        continue
                    # First non-comment line = header
                    if header is None:
                        header = stripped
                        out_f.write(line)
                    # Skip duplicate headers, write data
                    elif stripped != header:
                        out_f.write(line)
    print(f"TSV  → {out}")

def merge_gffs(f1, f2, out):
    """Merges GFF3 files, ensures ##gff-version appears only once at top."""
    Path(out).parent.mkdir(parents=True, exist_ok=True)
    version_written = False
    with open(out, "w") as out_f:
        for src in (f1, f2):
            with open(src, "r") as in_f:
                for line in in_f:
                    stripped = line.strip()
                    if stripped.startswith("##gff-version"):
                        if not version_written:
                            out_f.write(line)
                            version_written = True
                        continue
                    out_f.write(line)
    print(f"GFF  → {out}")


SEROTYPES = [1, 2, 3, 4]

for sero in SEROTYPES:
    print(f"\n🔹 Processing DENV-{sero}...")
    merge_fastas(
        CROP_SEQS_COL.format(sero),
        CROP_SEQS_EXP.format(sero),
        CROP_SEQS_MERGE.format(sero)
    )
    merge_tsvs(
        META_CLEAN_COL.format(sero),
        META_EXP.format(sero),
        META_MERGE.format(sero)
    )
    merge_gffs(
        TMP_GFF_COL.format(sero),
        TMP_GFF_EXP.format(sero),
        TMP_GFF_MERGE.format(sero)
    )

print("All merges complete.")
```

# Antigenic Space Projection

```python
import numpy as np
import pandas as pd
from Bio import SeqIO

# Global cache to store calculated Grantham distances
grantham_cache = {}

def get_grantham_dist(seq1, seq2):
    # Ensure symmetry to halve cache size (dist(A,B) == dist(B,A))
    if seq1 > seq2:
        seq1, seq2 = seq2, seq1
    key = (seq1, seq2)
    if key not in grantham_cache:
        grantham_cache[key] = grantham_distance(seq1, seq2)
    return grantham_cache[key]

all_exp_records = []
for sero in [1, 2, 3, 4]:
    records = [r for r in SeqIO.parse(AA_SEQS_EXP_E.format(sero), "fasta") 
               if not r.id.startswith("NODE")]
    all_exp_records.extend(records)

all_exp_ids = [r.id for r in all_exp_records]
all_exp_seqs = [str(r.seq[296:397]) for r in all_exp_records]  # EDIII region
n_all_exp = len(all_exp_seqs)

colombian_grantham_matrices = []

for sero in [1, 2, 3, 4]:
    print(f"Computing Grantham distances for DENV{sero}...")
    
    col_records = [r for r in SeqIO.parse(AA_SEQS_COL_E.format(sero), "fasta") 
                   if not r.id.startswith("NODE")]
    col_ids = [r.id for r in col_records]
    col_seqs = [str(r.seq[296:397]) for r in col_records]  # EDIII region
    
    n_col = len(col_seqs)
    matrix = np.zeros((n_col, n_all_exp))
    
    for i in range(n_col):
        for j in range(n_all_exp):
            dist, sites = get_grantham_dist(col_seqs[i], all_exp_seqs[j])
            matrix[i, j] = dist if sites > 0 else np.nan
    
    df_matrix = pd.DataFrame(matrix, index=col_ids, columns=all_exp_ids)
    colombian_grantham_matrices.append(df_matrix)
    
    print(f"  Matrix shape: {df_matrix.shape}")
    print(f"  Colombian strains: {len(col_ids)}")
    print(f"  Unified Experimental references: {len(all_exp_ids)}")
```

```python
from scipy.stats import spearmanr

mapa_ag_indexed = mapa_ag.drop_duplicates(subset='strainID').set_index('strainID')
projected_coordinates = {}
validation_results = []
projection_summary = []

for sero in [1, 2, 3, 4]:
    print(f"DENV{sero}:")
    grantham_matrix = colombian_grantham_matrices[sero - 1]
    colombian_strains = grantham_matrix.index.tolist()
    reference_ids = grantham_matrix.columns.tolist()

    # Exclude references without complete experimental coordinates.
    coords = mapa_ag_indexed.reindex(reference_ids)[['coord1', 'coord2', 'coord3']]
    valid_reference_mask = coords.notna().all(axis=1).to_numpy()
    coords = coords.loc[valid_reference_mask]
    ref_ids = coords.index.tolist()
    ref_coords = coords.to_numpy(dtype=float)

    D_all = grantham_matrix.loc[:, ref_ids].to_numpy(dtype=float)
    n_missing_distances = int(np.isnan(D_all).sum())
    n_distance_values = int(D_all.size)
    missing_distance_pct = 100 * n_missing_distances / n_distance_values if n_distance_values else np.nan
    projected = np.zeros((len(colombian_strains), 3))

    for i, strain_id in enumerate(colombian_strains):
        d_i = D_all[i, :].copy()
        finite_distances = np.isfinite(d_i)
        if not finite_distances.any():
            raise ValueError(f'No finite molecular distances available for {strain_id}')
        d_i[~finite_distances] = np.nanmax(d_i[finite_distances])

        weights = 1.0 / (d_i + 1e-6)
        weights /= weights.sum()
        a_init = (weights[:, np.newaxis] * ref_coords).sum(axis=0)
        max_d = np.max(d_i)
        d_norm = d_i / max_d if max_d > 0 else d_i

        def residuals(a):
            euclidean_dists = np.linalg.norm(ref_coords - a, axis=1)
            return weights * (euclidean_dists - d_norm)

        try:
            result = least_squares(
                residuals, a_init, method='lm', max_nfev=100,
                ftol=1e-8, xtol=1e-8, gtol=1e-8
            )
            projected[i] = result.x
            optimization_status = 'converged' if result.success else 'not_converged'
        except Exception:
            projected[i] = a_init
            optimization_status = 'initial_guess'

        fitted_distances = np.linalg.norm(ref_coords - projected[i], axis=1)
        residual = fitted_distances - d_norm
        spearman_rho = spearmanr(d_i, fitted_distances).statistic if len(d_i) > 1 else np.nan
        validation_results.append({
            'serotype': f'DENV{sero}',
            'strain': strain_id,
            'n_references': len(ref_ids),
            'n_missing_distances': int((~finite_distances).sum()),
            'missing_distance_pct': 100 * (~finite_distances).sum() / len(d_i),
            'rmse_normalized': float(np.sqrt(np.mean(residual ** 2))),
            'mae_normalized': float(np.mean(np.abs(residual))),
            'spearman_molecular_vs_fitted': float(spearman_rho),
            'optimization_status': optimization_status,
        })

    projected_coordinates[sero] = pd.DataFrame(
        projected, index=colombian_strains, columns=['X', 'Y', 'Z']
    )
    projection_summary.append({
        'serotype': f'DENV{sero}',
        'n_colombian_isolates': len(colombian_strains),
        'n_reference_strains': len(ref_ids),
        'n_reference_coordinates_missing': int((~valid_reference_mask).sum()),
        'reference_coordinate_missing_pct': 100 * (~valid_reference_mask).sum() / len(reference_ids),
        'n_molecular_distances_missing': n_missing_distances,
        'molecular_distance_missing_pct': missing_distance_pct,
    })

projection_summary = pd.DataFrame(projection_summary)
validation_results = pd.DataFrame(validation_results)
print('\nSummary by serotype:')
display(projection_summary)
print('\nProjection validation:')
display(validation_results.head())
```

```python
def run_projection_sensitivity(distance_matrices, metric_name, dimensions=(2, 3), reference_counts=(5, 10, None)):
    records = []
    for sero in [1, 2, 3, 4]:
        matrix = distance_matrices[sero - 1]
        reference_ids = matrix.columns.tolist()
        coords = mapa_ag_indexed.reindex(reference_ids)[['coord1', 'coord2', 'coord3']]
        valid_mask = coords.notna().all(axis=1).to_numpy()
        coords = coords.loc[valid_mask]
        ref_ids = coords.index.tolist()
        ref_coords_full = coords.to_numpy(dtype=float)
        D_full = matrix.loc[:, ref_ids].to_numpy(dtype=float)

        for dimension in dimensions:
            ref_coords = ref_coords_full[:, :dimension]
            for requested_count in reference_counts:
                rmses = []
                correlations = []
                n_refs = len(ref_ids) if requested_count is None else min(requested_count, len(ref_ids))
                for row in D_full:
                    finite = np.isfinite(row)
                    if not finite.any() or n_refs < dimension + 1:
                        continue
                    row = row.copy()
                    row[~finite] = np.nanmax(row[finite])
                    selected = np.argsort(row)[:n_refs]
                    d_i = row[selected]
                    coords_i = ref_coords[selected]
                    weights = 1.0 / (d_i + 1e-6)
                    weights /= weights.sum()
                    initial = (weights[:, np.newaxis] * coords_i).sum(axis=0)
                    d_norm = d_i / np.max(d_i) if np.max(d_i) > 0 else d_i

                    def residuals(point):
                        fitted = np.linalg.norm(coords_i - point, axis=1)
                        return weights * (fitted - d_norm)

                    result = least_squares(residuals, initial, method='lm', max_nfev=100)
                    fitted = np.linalg.norm(coords_i - result.x, axis=1)
                    rmses.append(np.sqrt(np.mean((fitted - d_norm) ** 2)))
                    correlations.append(spearmanr(d_i, fitted).statistic)

                records.append({
                    'metric': metric_name,
                    'serotype': f'DENV{sero}',
                    'dimensions': dimension,
                    'n_references': n_refs,
                    'n_isolates': len(rmses),
                    'mean_rmse_normalized': np.mean(rmses) if rmses else np.nan,
                    'mean_spearman': np.nanmean(correlations) if correlations else np.nan,
                })
    return pd.DataFrame(records)

metric_matrices = {'Grantham': colombian_grantham_matrices}
if 'colombian_hamming_matrices' in globals():
    metric_matrices['Hamming'] = colombian_hamming_matrices

sensitivity_results = pd.concat(
    [run_projection_sensitivity(matrices, name) for name, matrices in metric_matrices.items()],
    ignore_index=True
)
display(sensitivity_results)
if len(metric_matrices) == 1:
    print('Sensitivity to alternative metrics was not run: define colombian_hamming_matrices to add Hamming distances.')

# Optional validation against neutralization titers. The input table must contain
# verified antigen/serum identifiers shared with the reference map.
def validate_against_neutralization(titer_pairs, projected_distance_column='projected_distance'):
    required = {'antigen', 'serum', 'titer', projected_distance_column}
    missing = required.difference(titer_pairs.columns)
    if missing:
        raise ValueError(f'Missing columns for titer validation: {sorted(missing)}')
    valid = titer_pairs.dropna(subset=['titer', projected_distance_column]).copy()
    valid['log2_titer'] = np.log2(valid['titer'].astype(float))
    result = spearmanr(valid[projected_distance_column], -valid['log2_titer'])
    return {
        'n_pairs': len(valid),
        'spearman_projected_vs_log2_titer_loss': result.statistic,
        'p_value': result.pvalue,
    }

print('Titer validation requires a verified long-format table with antigen and serum IDs matched to the reference map.')
```

```python
mapa_ag_indexed = mapa_ag.drop_duplicates(subset='strainID').set_index('strainID')
ref_coords = mapa_ag_indexed.reindex(all_exp_ids)[['coord1', 'coord2', 'coord3']].fillna(0).values

projected_coordinates = {}
validation_results = {}

for sero in [1, 2, 3, 4]:
    print(f"DENV{sero}:")
    
    grantham_matrix = colombian_grantham_matrices[sero - 1]
    colombian_strains = grantham_matrix.index.tolist()
    
    D = grantham_matrix.values.astype(float)
    
    for i in range(D.shape[0]):
        nan_mask = np.isnan(D[i, :])
        if nan_mask.all():
            D[i, :] = 1.0
        else:
            D[i, nan_mask] = np.nanmax(D[i, :])
            
    projected = np.zeros((len(colombian_strains), 3))
    
    for i in range(len(colombian_strains)):
        d_i = D[i, :]
        weights = 1.0 / (d_i + 1e-6)
        weights = weights / weights.sum()
        a_init = (weights[:, np.newaxis] * ref_coords).sum(axis=0)
        
        def residuals(a):
            euclidean_dists = np.linalg.norm(ref_coords - a, axis=1)
            max_d = np.max(d_i)
            d_norm = d_i / max_d if max_d > 0 else d_i
            return weights * (euclidean_dists - d_norm)
        
        try:
            result = least_squares(residuals, a_init, method='lm', max_nfev=100,
                                 ftol=1e-8, xtol=1e-8, gtol=1e-8)
            projected[i] = result.x
        except Exception as e:
            print(f"    Warning: Optimization failed for {colombian_strains[i]}. Using initial guess.")
            projected[i] = a_init
    
    df_proj = pd.DataFrame(projected, index=colombian_strains, columns=['X', 'Y', 'Z'])
    projected_coordinates[sero] = df_proj
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D

unique_sero = sorted(list(projected_coordinates.keys()))
palette = sns.color_palette("husl", n_colors=len(unique_sero))
sero_colors = {sero: palette[i] for i, sero in enumerate(unique_sero)}

fig = plt.figure(figsize=(12, 10))

axes_info = [
    (1, 1, 1, 20, 45, "Perspective View")
]

axes = []
for rows, cols, index, elev, azim, title in axes_info:
    ax = fig.add_subplot(rows, cols, index, projection='3d')
    ax.view_init(elev=elev, azim=azim)
    ax.set_title(title, fontsize=14, fontweight='bold', pad=10)
    ax.set_xlabel('X', fontsize=11, labelpad=10)
    ax.set_ylabel('Y', fontsize=11, labelpad=10)
    ax.set_zlabel('Z', fontsize=11, labelpad=10)
    axes.append(ax)

for ax in axes:
    for sero in unique_sero:
        proj_coords = projected_coordinates[sero]
        color = sero_colors[sero]

        ax.scatter(proj_coords['X'], proj_coords['Y'], proj_coords['Z'],
                   c=[color], s=60, alpha=0.85,
                   marker='o', edgecolors='black', linewidth=0.4,
                   depthshade=True)

legend_elements = []
for sero in unique_sero:
    color = sero_colors[sero]
    n_samples = len(projected_coordinates[sero])
    
    legend_elements.append(Line2D([0], [0], marker='o', color='w', markerfacecolor=color, 
                                  markeredgecolor='black', markersize=8, 
                                  label=f'DENV-{sero} COL (n={n_samples})'))

axes[0].legend(handles=legend_elements, fontsize=9, loc='upper left', framealpha=0.9)

plt.tight_layout()
plt.savefig('../figures/colombian_projection_3d_perspective.png', dpi=300, bbox_inches='tight')
plt.show()
print("Zoomed 3D projection visualization (Perspective view, Colombian samples only) saved.")
```
